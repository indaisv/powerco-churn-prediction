```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ── Load the data ──────────────────────────────────────────
client_df = pd.read_csv('client_data.csv')
price_df  = pd.read_csv('price_data.csv')

# ── 1. Shape of data ───────────────────────────────────────
print("Client data shape:", client_df.shape)
print("Price data shape:", price_df.shape)
```

```python
# ── 2. Data types ──────────────────────────────────────────
print("CLIENT DATA TYPES")
print(client_df.dtypes)
print("\nPRICE DATA TYPES")
print(price_df.dtypes)
```

```python
# ── 3. Descriptive statistics ──────────────────────────────
print("CLIENT DATA STATISTICS")
client_df.describe()
```

```python
# ── 4. Missing values ──────────────────────────────────────
print("MISSING VALUES - CLIENT DATA")
print(client_df.isnull().sum())
print("\nMISSING VALUES - PRICE DATA")
print(price_df.isnull().sum())
```

```python
# ── 5. Churn distribution ──────────────────────────────────
client_df['churn'].value_counts().plot(kind='bar', color=['steelblue','tomato'])
plt.title('Churn Distribution')
plt.xlabel('Churned (1 = Yes, 0 = No)')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.show()
```

```python
# ── 6. Electricity consumption distribution ────────────────
sns.histplot(client_df['cons_12m'].dropna(), bins=50, color='steelblue')
plt.title('Distribution of Electricity Consumption (12 months)')
plt.xlabel('Consumption')
plt.ylabel('Count')
plt.show()
```

```python
# ── 7. Gas consumption distribution ───────────────────────
sns.histplot(client_df['cons_gas_12m'].dropna(), bins=50, color='green')
plt.title('Distribution of Gas Consumption (12 months)')
plt.xlabel('Consumption')
plt.ylabel('Count')
plt.show()
```

```python
# ── 8. Net margin distribution ─────────────────────────────
sns.histplot(client_df['net_margin'].dropna(), bins=50, color='purple')
plt.title('Distribution of Net Margin')
plt.xlabel('Net Margin')
plt.ylabel('Count')
plt.show()
```

```python
# ── 9. Price over time ─────────────────────────────────────
price_df['price_date'] = pd.to_datetime(price_df['price_date'])

price_df.groupby('price_date')['price_off_peak_var'].mean().plot()
plt.title('Average Off-Peak Energy Price Over Time')
plt.xlabel('Date')
plt.ylabel('Price')
plt.show()
```

```python
# ── 10. Churn rate by gas vs non-gas customers ─────────────
client_df.groupby('has_gas')['churn'].mean().plot(kind='bar', color=['steelblue','tomato'])
plt.title('Churn Rate - Gas vs Non-Gas Customers')
plt.xlabel('Has Gas (1 = Yes, 0 = No)')
plt.ylabel('Churn Rate')
plt.xticks(rotation=0)
plt.show()
```
